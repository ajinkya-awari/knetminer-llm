```python
from datetime import datetime, timezone
from pathlib import Path
import hashlib
import json
import urllib.request

APPROVAL = 'NOT_APPROVED'  # Change manually to APPROVED_PROJECT_06_LIVE_PREFLIGHT only after explicit approval.
WORK_ROOT = Path('/kaggle/working/06-knetminer-llm')
EVIDENCE_ROOT = WORK_ROOT / 'results' / 'kaggle_evidence'
EVIDENCE_ROOT.mkdir(parents=True, exist_ok=True)
timestamp = datetime.now(timezone.utc)
evidence_path = EVIDENCE_ROOT / f'live_preflight_{timestamp.strftime("%Y%m%dT%H%M%SZ")}.json'
endpoint = 'https://api.platform.opentargets.org/api/v4/graphql'
body = {'query': 'query Metadata { meta { dataVersion { year month } } }', 'variables': {}}
request_bytes = json.dumps({'method': 'POST', 'url': endpoint, 'body': body}, sort_keys=True, separators=(',', ':')).encode('utf-8')
record = {'status': 'blocked', 'stage': 'bounded_live_preflight', 'endpoint': endpoint, 'request_hash': hashlib.sha256(request_bytes).hexdigest(), 'timestamp_utc': timestamp.isoformat()}
print('STATUS: bounded Open Targets metadata preflight; no data query or model access.')
if APPROVAL != 'APPROVED_PROJECT_06_LIVE_PREFLIGHT':
    evidence_path.write_text(json.dumps(record, indent=2) + '\n', encoding='utf-8')
    print('BLOCKED: explicit approval comment is required before the provider call.')
    print(f'EVIDENCE: {evidence_path}')
    print('NEXT ACTION: obtain explicit approval, then edit only APPROVAL and rerun this cell manually.')
    raise SystemExit(1)
request = urllib.request.Request(endpoint, data=json.dumps(body).encode('utf-8'), headers={'Content-Type': 'application/json'}, method='POST')
try:
    with urllib.request.urlopen(request, timeout=10) as response:
        response_bytes = response.read()
    payload = json.loads(response_bytes)
    if not isinstance(payload, dict) or payload.get('errors'):
        raise ValueError('Open Targets returned an invalid GraphQL response')
    data = payload.get('data', {})
    version = data.get('meta', {}).get('dataVersion', {})
    release = f'{version.get("year")}.{version.get("month")}'
    record.update({'status': 'passed' if release == '26.06' else 'failed', 'release': release, 'response_hash': hashlib.sha256(response_bytes).hexdigest(), 'failure_category': None if release == '26.06' else 'release_mismatch'})
except Exception:
    record.update({'status': 'failed', 'failure_category': 'preflight_request_failed'})
evidence_path.write_text(json.dumps(record, indent=2) + '\n', encoding='utf-8')
print(f'EVIDENCE: {evidence_path}')
if record['status'] != 'passed':
    print('FAIL: preflight did not verify release 26.06; no later gate may run.')
    print('NEXT ACTION: preserve sanitized evidence and stop.')
    raise SystemExit(1)
print('PASS: release metadata preflight verified 26.06; no data was staged.')
print('NEXT ACTION: stop and request approval for the next gate.')
```
